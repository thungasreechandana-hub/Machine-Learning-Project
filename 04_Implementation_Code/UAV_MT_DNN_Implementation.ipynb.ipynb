```python
import os
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

import tensorflow as tf

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler, LabelEncoder
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    classification_report,
    confusion_matrix
)

from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestClassifier
from xgboost import XGBClassifier

print("TensorFlow version:", tf.__version__)
```

```python
# ================================================================
# MULTI-TASK DEEP NEURAL NETWORK FOR EMERGENCY MESSAGE PRIORITY
# AND SEMANTIC-LOSS TOLERANCE PREDICTION IN UAV NETWORKS
# ================================================================

# ------------------------------------------------
# 1. INSTALL REQUIRED LIBRARIES
# ------------------------------------------------
!pip -q install tensorflow scikit-learn xgboost

# ------------------------------------------------
# 2. IMPORT LIBRARIES
# ------------------------------------------------
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import tensorflow as tf

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler, LabelEncoder
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    classification_report,
    confusion_matrix
)

from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestClassifier
from xgboost import XGBClassifier

from tensorflow.keras.models import Model
from tensorflow.keras.layers import (
    Input,
    Dense,
    Dropout
)
from tensorflow.keras.optimizers import Adam

# ------------------------------------------------
# 3. CHECK TENSORFLOW
# ------------------------------------------------
print("TensorFlow version:", tf.__version__)

# ------------------------------------------------
# 4. LOAD FANET RESEARCH DATASET
# ------------------------------------------------
file_path = (
    "/content/FANET_Dataset/"
    "FANET_Dataset_NS3.40/"
    "Example_of_ML_ready/"
    "Link_Stability_For_ML_Scenario1.csv"
)

df = pd.read_csv(file_path)

print("\n================ DATASET INFORMATION ================")
print("Dataset shape:", df.shape)

print("\nColumns:")
print(df.columns.tolist())

print("\nMissing values:")
print(df.isnull().sum().sum())

# ------------------------------------------------
# 5. CREATE PROJECT-SPECIFIC LABELS
# ------------------------------------------------
# The original dataset contains link_lifetime_s and SNR_dB.
#
# Priority:
# Short link lifetime  -> High priority
# Medium lifetime      -> Medium priority
# Long lifetime        -> Low priority
#
# Semantic-loss tolerance:
# Low SNR  -> Low tolerance
# High SNR -> High tolerance

q25 = df["link_lifetime_s"].quantile(0.25)
q75 = df["link_lifetime_s"].quantile(0.75)

df["priority"] = pd.cut(
    df["link_lifetime_s"],
    bins=[-np.inf, q25, q75, np.inf],
    labels=["High", "Medium", "Low"]
)

snr_median = df["SNR_dB"].median()

df["semantic_loss"] = np.where(
    df["SNR_dB"] < snr_median,
    "Low",
    "High"
)

print("\n================ LABEL INFORMATION ================")

print("\nPriority distribution:")
print(df["priority"].value_counts())

print("\nSemantic-loss tolerance distribution:")
print(df["semantic_loss"].value_counts())

# ------------------------------------------------
# 6. SELECT INPUT FEATURES
# ------------------------------------------------
# We intentionally do NOT use:
# link_lifetime_s -> used to create priority label
# SNR_dB         -> used to create semantic-loss label
#
# This avoids direct target leakage.

features = [
    "distance_m",
    "nb_hello",
    "symmetrical_link",
    "new_link",
    "vel_x",
    "vel_y",
    "vel_z",
    "speed_m_s",
    "remaining_energy",
    "DataRate_Mbps"
]

X = df[features].copy()

# Remove any rows with undefined labels
valid = df["priority"].notna() & df["semantic_loss"].notna()

X = X.loc[valid]
df_valid = df.loc[valid].copy()

# ------------------------------------------------
# 7. ENCODE OUTPUT LABELS
# ------------------------------------------------
priority_encoder = LabelEncoder()
semantic_encoder = LabelEncoder()

y_priority = priority_encoder.fit_transform(
    df_valid["priority"].astype(str)
)

y_semantic = semantic_encoder.fit_transform(
    df_valid["semantic_loss"].astype(str)
)

print("\nPriority classes:")
print(priority_encoder.classes_)

print("\nSemantic-loss classes:")
print(semantic_encoder.classes_)

# ------------------------------------------------
# 8. TRAIN / TEST SPLIT
# ------------------------------------------------
# Combined labels are used for stratification
# so that both tasks remain represented.

combined_labels = (
    df_valid["priority"].astype(str)
    + "_"
    + df_valid["semantic_loss"].astype(str)
)

X_train, X_test, y_priority_train, y_priority_test, \
y_semantic_train, y_semantic_test = train_test_split(
    X,
    y_priority,
    y_semantic,
    test_size=0.20,
    random_state=42,
    stratify=combined_labels
)

print("\n================ DATA SPLIT ================")
print("Training samples:", len(X_train))
print("Testing samples :", len(X_test))

# ------------------------------------------------
# 9. FEATURE SCALING
# ------------------------------------------------
scaler = StandardScaler()

X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)

# ------------------------------------------------
# 10. BUILD MULTI-TASK DEEP NEURAL NETWORK
# ------------------------------------------------

input_layer = Input(
    shape=(len(features),),
    name="UAV_Network_Input"
)

# Shared layers
shared = Dense(
    64,
    activation="relu",
    name="Shared_Dense_1"
)(input_layer)

shared = Dropout(
    0.20,
    name="Shared_Dropout"
)(shared)

shared = Dense(
    32,
    activation="relu",
    name="Shared_Dense_2"
)(shared)

# ------------------------------------------------
# TASK 1: MESSAGE PRIORITY
# ------------------------------------------------

priority_branch = Dense(
    16,
    activation="relu",
    name="Priority_Branch"
)(shared)

priority_output = Dense(
    3,
    activation="softmax",
    name="Priority_Output"
)(priority_branch)

# ------------------------------------------------
# TASK 2: SEMANTIC-LOSS TOLERANCE
# ------------------------------------------------

semantic_branch = Dense(
    16,
    activation="relu",
    name="Semantic_Branch"
)(shared)

semantic_output = Dense(
    2,
    activation="softmax",
    name="Semantic_Output"
)(semantic_branch)

# ------------------------------------------------
# CREATE MODEL
# ------------------------------------------------

model = Model(
    inputs=input_layer,
    outputs=[
        priority_output,
        semantic_output
    ]
)

# ------------------------------------------------
# 11. COMPILE MODEL
# ------------------------------------------------

model.compile(
    optimizer=Adam(learning_rate=0.001),

    loss={
        "Priority_Output":
            "sparse_categorical_crossentropy",

        "Semantic_Output":
            "sparse_categorical_crossentropy"
    },

    metrics={
        "Priority_Output": ["accuracy"],
        "Semantic_Output": ["accuracy"]
    }
)

print("\n================ MODEL ARCHITECTURE ================")
model.summary()

# ------------------------------------------------
# 12. TRAIN MT-DNN
# ------------------------------------------------

print("\n================ TRAINING MT-DNN ================")

history = model.fit(
    X_train_scaled,

    {
        "Priority_Output": y_priority_train,
        "Semantic_Output": y_semantic_train
    },

    epochs=30,
    batch_size=32,

    validation_split=0.20,

    verbose=1
)

# ------------------------------------------------
# 13. PREDICTIONS
# ------------------------------------------------

priority_pred_prob, semantic_pred_prob = model.predict(
    X_test_scaled
)

priority_pred = np.argmax(
    priority_pred_prob,
    axis=1
)

semantic_pred = np.argmax(
    semantic_pred_prob,
    axis=1
)

# ------------------------------------------------
# 14. FUNCTION FOR METRICS
# ------------------------------------------------

def calculate_metrics(y_true, y_pred):

    accuracy = accuracy_score(
        y_true,
        y_pred
    )

    precision = precision_score(
        y_true,
        y_pred,
        average="weighted",
        zero_division=0
    )

    recall = recall_score(
        y_true,
        y_pred,
        average="weighted",
        zero_division=0
    )

    f1 = f1_score(
        y_true,
        y_pred,
        average="weighted",
        zero_division=0
    )

    return accuracy, precision, recall, f1

# ------------------------------------------------
# 15. PRIORITY RESULTS
# ------------------------------------------------

priority_accuracy, priority_precision, \
priority_recall, priority_f1 = calculate_metrics(
    y_priority_test,
    priority_pred
)

# ------------------------------------------------
# 16. SEMANTIC RESULTS
# ------------------------------------------------

semantic_accuracy, semantic_precision, \
semantic_recall, semantic_f1 = calculate_metrics(
    y_semantic_test,
    semantic_pred
)

# ------------------------------------------------
# 17. DISPLAY RESULTS
# ------------------------------------------------

print("\n\n================================================")
print("             MT-DNN FINAL RESULTS")
print("================================================")

print("\nMESSAGE PRIORITY CLASSIFICATION")
print("--------------------------------")
print(f"Accuracy  : {priority_accuracy:.4f}")
print(f"Precision : {priority_precision:.4f}")
print(f"Recall    : {priority_recall:.4f}")
print(f"F1 Score  : {priority_f1:.4f}")

print("\nSEMANTIC-LOSS TOLERANCE CLASSIFICATION")
print("----------------------------------------")
print(f"Accuracy  : {semantic_accuracy:.4f}")
print(f"Precision : {semantic_precision:.4f}")
print(f"Recall    : {semantic_recall:.4f}")
print(f"F1 Score  : {semantic_f1:.4f}")

# ------------------------------------------------
# 18. CLASSIFICATION REPORT - PRIORITY
# ------------------------------------------------

print("\n================ PRIORITY CLASSIFICATION REPORT ================")

print(
    classification_report(
        y_priority_test,
        priority_pred,
        target_names=priority_encoder.classes_,
        zero_division=0
    )
)

# ------------------------------------------------
# 19. CLASSIFICATION REPORT - SEMANTIC LOSS
# ------------------------------------------------

print("\n================ SEMANTIC-LOSS CLASSIFICATION REPORT ================")

print(
    classification_report(
        y_semantic_test,
        semantic_pred,
        target_names=semantic_encoder.classes_,
        zero_division=0
    )
)

# ------------------------------------------------
# 20. CONFUSION MATRIX - PRIORITY
# ------------------------------------------------

cm_priority = confusion_matrix(
    y_priority_test,
    priority_pred
)

plt.figure(figsize=(6, 5))

sns.heatmap(
    cm_priority,
    annot=True,
    fmt="d",
    cmap="Blues",
    xticklabels=priority_encoder.classes_,
    yticklabels=priority_encoder.classes_
)

plt.xlabel("Predicted")
plt.ylabel("Actual")
plt.title("MT-DNN - Message Priority Confusion Matrix")
plt.show()

# ------------------------------------------------
# 21. CONFUSION MATRIX - SEMANTIC LOSS
# ------------------------------------------------

cm_semantic = confusion_matrix(
    y_semantic_test,
    semantic_pred
)

plt.figure(figsize=(6, 5))

sns.heatmap(
    cm_semantic,
    annot=True,
    fmt="d",
    cmap="Greens",
    xticklabels=semantic_encoder.classes_,
    yticklabels=semantic_encoder.classes_
)

plt.xlabel("Predicted")
plt.ylabel("Actual")
plt.title("MT-DNN - Semantic-Loss Tolerance Confusion Matrix")
plt.show()

# ------------------------------------------------
# 22. TRAINING ACCURACY
# ------------------------------------------------

plt.figure(figsize=(8, 5))

plt.plot(
    history.history["Priority_Output_accuracy"],
    label="Priority Training Accuracy"
)

plt.plot(
    history.history["val_Priority_Output_accuracy"],
    label="Priority Validation Accuracy"
)

plt.plot(
    history.history["Semantic_Output_accuracy"],
    label="Semantic Training Accuracy"
)

plt.plot(
    history.history["val_Semantic_Output_accuracy"],
    label="Semantic Validation Accuracy"
)

plt.xlabel("Epoch")
plt.ylabel("Accuracy")
plt.title("MT-DNN Training and Validation Accuracy")
plt.legend()
plt.grid(True)
plt.show()

# ------------------------------------------------
# 23. BASELINE MODEL 1
# LOGISTIC REGRESSION
# ------------------------------------------------

logistic_model = LogisticRegression(
    max_iter=1000
)

logistic_model.fit(
    X_train_scaled,
    y_priority_train
)

logistic_pred = logistic_model.predict(
    X_test_scaled
)

logistic_accuracy = accuracy_score(
    y_priority_test,
    logistic_pred
)

# ------------------------------------------------
# 24. BASELINE MODEL 2
# RANDOM FOREST
# ------------------------------------------------

rf_model = RandomForestClassifier(
    n_estimators=100,
    random_state=42
)

rf_model.fit(
    X_train,
    y_priority_train
)

rf_pred = rf_model.predict(
    X_test
)

rf_accuracy = accuracy_score(
    y_priority_test,
    rf_pred
)

# ------------------------------------------------
# 25. BASELINE MODEL 3
# XGBOOST
# ------------------------------------------------

xgb_model = XGBClassifier(
    n_estimators=100,
    max_depth=5,
    learning_rate=0.1,
    random_state=42,
    eval_metric="mlogloss"
)

xgb_model.fit(
    X_train,
    y_priority_train
)

xgb_pred = xgb_model.predict(
    X_test
)

xgb_accuracy = accuracy_score(
    y_priority_test,
    xgb_pred
)

# ------------------------------------------------
# 26. MODEL COMPARISON
# ------------------------------------------------

comparison = pd.DataFrame({
    "Model": [
        "MT-DNN",
        "Logistic Regression",
        "Random Forest",
        "XGBoost"
    ],

    "Priority Accuracy": [
        priority_accuracy,
        logistic_accuracy,
        rf_accuracy,
        xgb_accuracy
    ]
})

print("\n================ MODEL COMPARISON ================")

print(
    comparison.to_string(
        index=False
    )
)

# ------------------------------------------------
# 27. MODEL COMPARISON GRAPH
# ------------------------------------------------

plt.figure(figsize=(9, 5))

plt.bar(
    comparison["Model"],
    comparison["Priority Accuracy"]
)

plt.ylabel("Accuracy")
plt.xlabel("Model")
plt.title("Priority Classification - Model Comparison")

plt.ylim(
    0,
    1
)

plt.xticks(
    rotation=15
)

plt.grid(
    axis="y"
)

plt.show()

# ------------------------------------------------
# 28. SAMPLE UAV MESSAGE PREDICTION
# ------------------------------------------------

sample = X_test.iloc[[0]]

sample_scaled = scaler.transform(
    sample
)

sample_priority_prob, \
sample_semantic_prob = model.predict(
    sample_scaled,
    verbose=0
)

sample_priority_index = np.argmax(
    sample_priority_prob[0]
)

sample_semantic_index = np.argmax(
    sample_semantic_prob[0]
)

sample_priority = priority_encoder.inverse_transform(
    [sample_priority_index]
)[0]

sample_semantic = semantic_encoder.inverse_transform(
    [sample_semantic_index]
)[0]

print("\n================ SAMPLE UAV PREDICTION ================")

print("Input network features:")

print(
    sample.to_string(
        index=False
    )
)

print("\nPredicted Message Priority:")
print(sample_priority)

print("\nPredicted Semantic-Loss Tolerance:")
print(sample_semantic)

# ------------------------------------------------
# 29. SAVE TRAINED MODEL
# ------------------------------------------------

model.save(
    "/content/UAV_MT_DNN_Model.keras"
)

print("\n================================================")
print("PROJECT EXECUTION COMPLETED SUCCESSFULLY")
print("================================================")

print("\nSaved model:")
print("/content/UAV_MT_DNN_Model.keras")
```
